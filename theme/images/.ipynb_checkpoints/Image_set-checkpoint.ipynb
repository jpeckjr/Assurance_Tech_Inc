```python
from PIL import Image

# Open the original image
img = Image.open("your-image.jpg")

# Resize (e.g., half the original size)
resized_img = img.resize((img.width // 2, img.height // 2))

# Save the resized image
resized_img.save("your-image-small.jpg")
```

```python
from PIL import Image, ImageEnhance

def apply_tint(input_path, output_path, tint_color=(0, 0, 0), opacity=0.5):
    # Open the original image
    image = Image.open(input_path).convert("RGBA")

    # Create a tint overlay
    overlay = Image.new("RGBA", image.size, tint_color + (int(255 * opacity),))

    # Apply the overlay
    tinted_image = Image.alpha_composite(image, overlay)

    # Save the result
    tinted_image.convert("RGB").save(output_path)

# Example usage
apply_tint("banner_team_new_fix.jpg", "banner_team_new_tinted.jpg", tint_color=(0, 0, 0), opacity=0.5)
```
